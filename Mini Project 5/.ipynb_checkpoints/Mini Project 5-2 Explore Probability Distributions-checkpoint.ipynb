## **Step 3: Statistical tests**

Use the empirical rule to observe the data, then test and verify that it is normally distributed.


 As you have learned, the empirical rule states that, for every normal distribution: 
- 68% of the data fall within 1 standard deviation of the mean
- 95% of the data fall within 2 standard deviations of the mean
- 99.7% of the data fall within 3 standard deviations of the mean


First, define two variables to store the mean and standard deviation, respectively, for `aqi_log`. Creating these variables will help you easily access these measures as you continue with the calculations involved in applying the empirical rule. 

```python
aqi_log_mean = np.mean(df['aqi_log'])
print("Mean of aqi_log:", aqi_log_mean)
```

```python
aqi_log_std = np.std(df['aqi_log'], ddof=1) 
print("Standard deviation of aqi_log:", aqi_log_std)
```

Now, check the first part of the empirical rule: whether 68% of the `aqi_log` data falls within 1 standard deviation of the mean.

To compute the actual percentage of the data that satisfies this criteria, define the lower limit (for example, 1 standard deviation below the mean) and the upper limit (for example, 1 standard deviation above the mean). This will enable you to create a range and confirm whether each value falls within it.

```python
lower_limit_1 = aqi_log_mean - aqi_log_std
upper_limit_1 = aqi_log_mean + aqi_log_std

within_1_std = df[(df['aqi_log'] >= lower_limit_1) & (df['aqi_log'] <= upper_limit_1)]
percent_within_1_std = len(within_1_std) / len(df) * 100

print(f"1 Standard Deviation Range: ({lower_limit_1}, {upper_limit_1})")
```

```python
# Display the actual percentage of data that falls within 1 standard deviation of the mean.
print(f"Percentage within 1 standard deviation: {percent_within_1_std:.2f}%")
```

Now, consider the second part of the empirical rule: whether 95% of the `aqi_log` data falls within 2 standard deviations of the mean.

To compute the actual percentage of the data that satisfies this criteria, define the lower limit (for example, 2 standard deviations below the mean) and the upper limit (for example, 2 standard deviations above the mean). This will enable you to create a range and confirm whether each value falls within it.

```python
lower_limit_2 = aqi_log_mean - 2 * aqi_log_std
upper_limit_2 = aqi_log_mean + 2 * aqi_log_std

within_2_std = df[(df['aqi_log'] >= lower_limit_2) & (df['aqi_log'] <= upper_limit_2)]
percent_within_2_std = len(within_2_std) / len(df) * 100

print(f"2 Standard Deviations Range: ({lower_limit_2}, {upper_limit_2})")
```

```python
# Display the actual percentage of data that falls within 2 standard deviations of the mean.
print(f"Percentage within 2 standard deviations: {percent_within_2_std:.2f}%")
```

Now, consider the third part of the empirical rule:whether 99.7% of the `aqi_log` data falls within 3 standard deviations of the mean.

To compute the actual percentage of the data that satisfies this criteria, define the lower limit (for example, 3 standard deviations below the mean) and the upper limit (for example, 3 standard deviations above the mean). This will enable you to create a range and confirm whether each value falls within it.

```python
lower_limit_3 = aqi_log_mean - 3 * aqi_log_std
upper_limit_3 = aqi_log_mean + 3 * aqi_log_std

within_3_std = df[(df['aqi_log'] >= lower_limit_3) & (df['aqi_log'] <= upper_limit_3)]
percent_within_3_std = len(within_3_std) / len(df) * 100

print(f"3 Standard Deviations Range: ({lower_limit_3}, {upper_limit_3})")
```

```python
# Display the actual percentage of data that falls within 3 standard deviations of the mean.

print(f"Percentage within 3 standard deviations: {percent_within_3_std:.2f}%")
```

**Question:** How would you use z-score to find outliers? 

A: look for high z-score

Compute the z-score for every `aqi_log` value. Then, add a column named `z_score` in the data to store those results. 

```python
df['z_score'] = (df['aqi_log'] - aqi_log_mean) / aqi_log_std
df.head()
```

Identify the parts of the data where `aqi_log` is above or below 3 standard deviations of the mean.

```python
outliers = df[(df['z_score'] > 3) | (df['z_score'] < -3)]
outliers
```
